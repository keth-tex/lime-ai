# Deployment Workflow: Backend (Lime-AI)

Dieses Notebook steuert den Synchronisations- und Deployment-Prozess des Python-Backends vom lokalen Entwicklungsrechner (Windows via WSL) auf den Netcup-Server.

Wichtig: Lokale Windows-Bibliotheken (venv) und SQLite-Datenbanken (*.db) werden automatisch ausgeschlossen.

### 1. Code-Dateien mit dem Server synchronisieren

Überträgt geänderte Python-Skripte und Konfigurationen via rsync. Plattformspezifische und volatile Dateien werden ignoriert.

```python
# Synchronisiert den Ordner 'backend' auf den Server
!wsl rsync -avz --delete --exclude 'venv/' --exclude '.venv/' --exclude 'pycache/' --exclude '.pyc' --exclude '.git/' --exclude '.env' --exclude '.db' --exclude 'data/' --exclude 'cache/' backend/ thomas@62.83.19.50:/home/thomas/lime-ai/backend/
```

### 2. Python-Abhängigkeiten aktualisieren (Optional)

Führe diese Zelle nur aus, wenn neue Pakete in die requirements.txt aufgenommen wurden.

```python
!wsl ssh thomas@62.83.19.50 "cd /home/thomas/lime-ai/backend && source venv/bin/activate && pip install -r requirements.txt"
```

### 3. Hintergrunddienst neu starten und Logs prüfen

Startet den Dienst lime-ai über Systemd neu und gibt unmittelbar die letzten 25 Zeilen der Systemlogs aus.

```python
!wsl ssh thomas@62.83.19.50 "sudo systemctl restart lime-ai && sleep 2 && sudo journalctl -u lime-ai -n 25 --no-pager"
```
