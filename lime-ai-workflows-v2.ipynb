Da dein VS Code unter Windows läuft, deine Werkzeuge (`rsync`, `ssh`) aber in WSL liegen, leiten wir die Befehle mit dem Präfix `!wsl` direkt an dein Linux-Subsystem weiter.

## 1. Lokales Python-Backend starten
Da das Backend nativ unter Windows läuft, wird auf die Ausführung in WSL verzichtet. Der Start erfolgt über das mitgelieferte Batch-Skript, welches den Server auf Port 8080 hochfährt.

**Wichtig:** Da der Server dauerhaft läuft, würde das Ausführen in einer Notebook-Zelle den Kernel blockieren. Starte das Backend stattdessen in einem separaten Windows-Terminal.
```cmd
cd backend

```

## 2. Lokalen Development-Server starten

Um UI-Anpassungen (Svelte-Komponenten, CSS, JS) während der Entwicklung zu testen.

**Wichtig:** Da der Server dauerhaft läuft, würde das Ausführen in einer Notebook-Zelle den Kernel blockieren. Starte das Backend stattdessen in einem separaten Windows-Terminal.

```cmd

```

## 3. Frontend kompilieren (Lokaler Build)
Führe diese Zelle aus, um das Svelte-Frontend nach lokalen Anpassungen am Code oder Design neu zu bauen. 
*(Hinweis: Falls du Node.js nicht in WSL, sondern direkt in Windows installiert hast, entferne das `wsl` aus dem Befehl.)*

````python
!wsl npm run build
````

## 4. Deployment: Frontend auf Netcup pushen
Diese Zelle nutzt `rsync` aus WSL heraus, um deinen lokalen `build/`-Ordner mit dem Server abzugleichen.

````python
!wsl rsync -avz --delete build/ thomas@62.83.19.50:/home/thomas/lime-ai/build/
````

## 5. Server-Monitoring (Remote via SSH)
Hier führen wir über WSL direkt SSH-Befehle auf dem Netcup-Server aus, um den Status abzufragen.

````python
# Prüft, ob Nginx fehlerfrei läuft
!wsl ssh thomas@62.83.19.50 "sudo systemctl status nginx --no-pager"
````

````python
# Gibt die letzten Log-Einträge des Python-Backends aus
!wsl ssh thomas@62.83.19.50 "sudo journalctl -u lime-ai -n 15 --no-pager"
````
